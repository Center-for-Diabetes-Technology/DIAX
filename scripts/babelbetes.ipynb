# BabelBetes Converter

This notebook documents how to convert BabelBetes parquet outputs into DIAX format.

[babelbetes](https://github.com/nudgebg/babelbetes) is another dataset standardization tool that operates on similar datasets to DIAX,
so we provide this utility to help translate between the two standards. 
The babelbetes repository is well maintained and documented. 
Some keys are present in DIAX that are not present in babelbetes, along with metadata information about device or insulin types. 

````python
import pandas as pd
import numpy as np
import os
import json
import datetime

data_source = '../../data_raw/babelbetes/data/out'
output_dir = '../../diax/babelbetes/'  # or just ../../diax/

if not os.path.exists(output_dir):
    os.makedirs(output_dir)
````

````python
import os
import sys

sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..', 'scripts')))
import babelbetes_converter as bbc

bbc.convert_all(data_source, output_dir, studies=None)
````
